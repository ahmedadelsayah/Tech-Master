## 6. Try text cleaning on a few samples
قبل ما تتطبق على الداتاست كله، جرب الدالة على شوية أمثلة وشوف الناتج.

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    text = " ".join(text.split())
    return text

for sample in df["message"].head(5):
    print("Before:", sample)
    print("After: ", clean_text(sample))
    print("-" * 60)
```

## 7. Most common words per class (quick look)
نظرة سريعة على الكلمات الأكتر تكرارًا في كل فئة — بيديك حدس عن الـ features المهمة
قبل ما تعمل TF-IDF فعليًا في `features.py`.

```python
from collections import Counter

def top_words(messages, n=15):
    words = " ".join(messages.apply(clean_text)).split()
    return Counter(words).most_common(n)

print("Top words in SPAM:")
print(top_words(df[df["label"] == "spam"]["message"]))

print("\nTop words in HAM:")
print(top_words(df[df["label"] == "ham"]["message"]))
```
